```python
import pandas as pd
from pathlib import Path

data_folder = Path("./data_raw")  # Path to the folder containing the CSV files

lables = {
    "1ct" : 0,
    "2ct" : 0,
    "5ct" : 0,
    "10ct" : 0,
    "20ct" : 0,
    "50ct" : 0,
    "1euro" : 0,
    "2euro" : 0
} 

for file in data_folder.iterdir():
    # Extract the first number from the filename
    label = int(file.name.split("_")[0])

    # Update the count for the corresponding label
    if label == 1:
        lables["1ct"] += 1
    elif label == 2:
        lables["2ct"] += 1
    elif label == 5:
        lables["5ct"] += 1
    elif label == 10:
        lables["10ct"] += 1
    elif label == 20:
        lables["20ct"] += 1
    elif label == 50:
        lables["50ct"] += 1
    elif label == 100:
        lables["1euro"] += 1
    elif label == 200:
        lables["2euro"] += 1

print("Label counts:")
print("----------------")
print(f"Total files: {sum(lables.values())}")
print("----------------")
for label, count in lables.items():
    print(f"{label}: {count}")
print("----------------")
```
